Question 28:Calculate compound interest using math.pow().

```python
principal = 1000
rate = 5
time = 2
amount = principal * math.pow((1 + rate / 100), time)
compound_interest = amount - principal
print("Amount =", amount)
print("Compound Interest =", compound_interest)
```

Question 29:Find the area and circumference of a circle using math.pi.

```python
radies =5
area = math.pi * radies**2
circumference = 2 * math.pi * radies
print("Area =", area)
print("Circumference =", circumference)
```

Question 30:Create a scientific calculator that supports square root, factorial, logarithm, power, trigonometric functions, GCD, and LCM.

```python
import math
choice = int(input("Enter your choice (1-7): "))

if choice == 1:
    n = float(input("Enter a number: "))
    print("Square Root =", math.sqrt(n))

elif choice == 2:
    n = int(input("Enter a number: "))
    print("Factorial =", math.factorial(n))

elif choice == 3:
    n = float(input("Enter a number: "))
    print("Log(base10) =", math.log10(n))

elif choice == 4:
    base = float(input("Enter base: "))
    exp = float(input("Enter exponent: "))
    print("Power =", math.pow(base, exp))

elif choice == 5:
    angle = float(input("Enter angle in degrees: "))
    rad = math.radians(angle)

    print("sin =", math.sin(rad))
    print("cos =", math.cos(rad))
    print("tan =", math.tan(rad))

elif choice == 6:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    print("GCD =", math.gcd(a, b))

elif choice == 7:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    print("LCM =", math.lcm(a, b))

else:
    print("Invalid Choice")
```
