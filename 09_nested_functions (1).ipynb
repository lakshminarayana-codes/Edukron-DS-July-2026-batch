```python
def analyze_text(string):
    Total_characters=len(string)
    Total_words=len(string.split())
    Total_vowels=0
    Total_digits=0
    Total_spaces=0
    for i in string:
        if i in "AEIOUaeiou":
            Total_vowels +=1
        elif i.isdigit():
            Total_digits +=1
        elif i.isspace():
            Total_spaces +=1
    return {"Total Characters": Total_characters,
        "Total Words": Total_words,
        "Total Vowels": Total_vowels,
        "Total Digits": Total_digits,
        "Total Spaces": Total_spaces
    }
string=input('Enter the string')
result=analyze_text(string)
for key,value in result.items():
    print(f'{key} : {value}')


```

Use filter().
A basic valid email should:
Contain exactly one @.
Have characters before @.
Contain a dot after @.

```python
def create_bank_account(balance):

    def deposit(amount):
        nonlocal balance
        balance += amount
        print(f"Deposited: {amount}")

    def withdraw(amount):
        nonlocal balance
        if amount <= balance:
            balance -= amount
            print(f"Withdrawn: {amount}")
        else:
            print("Insufficient balance")

    def check_balance():
        print("Current Balance:", balance)

    return deposit, withdraw, check_balance


deposit, withdraw, check_balance = create_bank_account(1000)

check_balance()
deposit(500)
check_balance()
withdraw(300)
check_balance()
withdraw(1500)
check_balance()
```

It should print another message after execution.
Apply it to an addition function.
Preserve the original function's return value.
Expected Pattern
Starting function...
Result: 30
Function completed.

```python
def execution_logger(func):

    def wrapper(a, b):
        print("Starting function...")
        result = func(a, b)
        print("Result:", result)
        print("Function completed.")
        return result

    return wrapper


@execution_logger
def addition(a, b):
    return a + b


addition(10, 20)
```

```python
# Global Variable (Global Scope)
company = "ABC Technologies"

employees = [
    {
        "id": 101,
        "name": "Ravi",
        "department": "IT",
        "salary": 60000,
        "performance_score": 88,
        "experience": 5
    },
    {
        "id": 102,
        "name": "Anitha",
        "department": "HR",
        "salary": 45000,
        "performance_score": 72,
        "experience": 4
    },
    {
        "id": 103,
        "name": "Kiran",
        "department": "IT",
        "salary": 75000,
        "performance_score": 94,
        "experience": 7
    },
    {
        "id": 104,
        "name": "Meena",
        "department": "Finance",
        "salary": 55000,
        "performance_score": 81,
        "experience": 6
    },
    {
        "id": 105,
        "name": "Arjun",
        "department": "IT",
        "salary": 40000,
        "performance_score": 65,
        "experience": 2
    }
]

# 1. Normal Function
def display_company():
    print("Company:", company)

# 2. Filter High Performers
def high_performers():
    return list(filter(lambda emp: emp["performance_score"] >= 80, employees))

# 3. Map Bonus
def calculate_bonus():
    return list(map(
        lambda emp: {
            "Name": emp["name"],
            "Bonus": emp["salary"] * 0.10
        },
        employees
    ))

# 4. Reduce Total Salary

def total_salary():
    salaries = list(map(lambda emp: emp["salary"], employees))
    return reduce(lambda x, y: x + y, salaries, 0)



# 5. Nested Function

def department_report():

    def report():
        it = list(filter(lambda e: e["department"] == "IT", employees))
        print("\nIT Employees")
        for emp in it:
            print(emp["name"])

    report()


# 6. Closure Example

def create_increment(percent):

    def increment(employee):
        new_salary = employee["salary"] + employee["salary"] * percent / 100
        return {
            "Name": employee["name"],
            "New Salary": new_salary
        }

    return increment



# 7. LEGB Example
message = "Global Message"

def outer():

    message = "Enclosing Message"

    def inner():

        message = "Local Message"

        print("\nLEGB Demonstration")
        print(message)

    inner()

# Main Program

display_company()

print("\nHigh Performers")
for emp in high_performers():
    print(emp["name"], emp["performance_score"])

print("\nEmployee Bonus")
for emp in calculate_bonus():
    print(emp)

print("\nTotal Salary:", total_salary())

department_report()

print("\nSalary Increment (20%)")
increment20 = create_increment(20)

for emp in employees:
    print(increment20(emp))

outer()
```
