```python
class InvalidTransactionError(Exception):
    pass

def validate_transaction(data):
    try:
        amount = float(data["amount"])
        if amount <= 0:
            raise ValueError("amount must be greater than zero")
    except (KeyError, TypeError, ValueError) as e:
        raise InvalidTransactionError("Invalid transaction for ACC-1032") from e
    return amount

try:
    validate_transaction({"amount": "bad"})
except InvalidTransactionError as e:
    print(e)
    print("Original cause:", type(e.__cause__).__name__, "-", e.__cause__)
```

### Q33. Process transfers while isolating bad rows for banking case 1; use sample reference `ACC-1033` and explain the result.

**Hints**

1. Identify the smallest risky statement and the narrow exception it can raise.
2. Keep business-rule failures distinct from programming or infrastructure failures.
3. For medium/difficult work, preserve the original cause and protect partially updated balances.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
accounts = {"A": 5000.0, "B": 1000.0, "C": 300.0}
transfers = [
    {"from": "A", "to": "B", "amount": 500},
    {"from": "A", "to": "C", "amount": "bad"},
    {"from": "X", "to": "B", "amount": 100}
]
successes = []
failures = []

for index, row in enumerate(transfers, start=1):
    try:
        source = row["from"]
        target = row["to"]
        amount = float(row["amount"])
        if source not in accounts or target not in accounts:
            raise ValueError("Unknown account")
        if amount <= 0:
            raise ValueError("Amount must be greater than zero")
        if accounts[source] < amount:
            raise ValueError("Insufficient funds")
        accounts[source] -= amount
        accounts[target] += amount
        successes.append(index)
    except (KeyError, TypeError, ValueError) as e:
        failures.append({"row": index, "error": str(e)})

print("Successful rows:", successes)
print("Failed rows:", failures)
print("Balances:", accounts)
```

### Q34. Retry a transient payment gateway for banking case 1; use sample reference `ACC-1034` and explain the result.

**Hints**

1. Identify the smallest risky statement and the narrow exception it can raise.
2. Keep business-rule failures distinct from programming or infrastructure failures.
3. For medium/difficult work, preserve the original cause and protect partially updated balances.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
class TransientGatewayError(Exception):
    pass

attempts = {"count": 0}

def payment_gateway():
    attempts["count"] += 1
    if attempts["count"] < 3:
        raise TransientGatewayError("Temporary gateway unavailable")
    return "PAYMENT_ACCEPTED"

max_retries = 3
result = "PAYMENT_FAILED"
for attempt in range(1, max_retries + 1):
    try:
        result = payment_gateway()
    except TransientGatewayError as e:
        print("Attempt", attempt, "failed:", e)
        if attempt == max_retries:
            break
    else:
        print("Payment succeeded on attempt", attempt)
        break

print("Result:", result)
```

### Q94. Implement bounded exponential retry for banking case 3; use sample reference `ACC-1094` and explain the result.

**Hints**

1. Identify the smallest risky statement and the narrow exception it can raise.
2. Keep business-rule failures distinct from programming or infrastructure failures.
3. For medium/difficult work, preserve the original cause and protect partially updated balances.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
import time

class TransientError(Exception):
    pass

attempts = 0
max_attempts = 4
base_delay = 0.05
max_delay = 0.2

def operation():
    global attempts
    attempts += 1
    if attempts < 3:
        raise TransientError("Temporary failure")
    return "SUCCESS"

result = "FAILED"
for attempt in range(1, max_attempts + 1):
    try:
        result = operation()
    except TransientError as e:
        print("Attempt", attempt, "failed:", e)
        if attempt == max_attempts:
            break
        delay = min(base_delay * (2 ** (attempt - 1)), max_delay)
        print("Retrying after", delay, "seconds")
        time.sleep(delay)
    else:
        break

print("Reference:", "ACC-1094")
print("Final result:", result)
```

### Q95. Create a transaction context manager for banking case 3; use sample reference `ACC-1095` and explain the result.

**Hints**

1. Identify the smallest risky statement and the narrow exception it can raise.
2. Keep business-rule failures distinct from programming or infrastructure failures.
3. For medium/difficult work, preserve the original cause and protect partially updated balances.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
class Transaction:
    def __init__(self, accounts):
        self.accounts = accounts
        self.original = accounts.copy()

    def __enter__(self):
        return self

    def transfer(self, source, target, amount):
        if source not in self.accounts or target not in self.accounts:
            raise KeyError("Unknown account")
        if amount <= 0:
            raise ValueError("Amount must be greater than zero")
        if self.accounts[source] < amount:
            raise ValueError("Insufficient funds")
        self.accounts[source] -= amount
        self.accounts[target] += amount

    def __exit__(self, exc_type, exc_value, traceback):
        if exc_type is not None:
            self.accounts.clear()
            self.accounts.update(self.original)
            print("Transaction rolled back")
            return False
        print("Transaction committed")
        return False

accounts = {"A": 5000.0, "B": 1000.0}
try:
    with Transaction(accounts) as tx:
        tx.transfer("A", "B", 750.0)
        raise RuntimeError("Simulated transaction failure")
except RuntimeError as e:
    print("Error:", e)

print("Reference:", "ACC-1095")
print("Balances:", accounts)
```
